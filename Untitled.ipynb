```python
"""
Power Method Devig Calculator

Given decimal odds (e.g., 1.91, 2.10), this script computes the "fair" implied probabilities
by raising each raw implied probability (1/odds) to a power t and scaling so they sum to 1.
The exponent t is chosen so that the sum of the powered values equals 1 directly, i.e.:

    sum_i (1/odds_i)^t = 1   -->   p_i = (1/odds_i)^t

This is a common "multiplicative power" method for removing bookmaker overround.
"""

import math

def power_method_devig(odds, tol=1e-12, max_iter=100):
    """
    Compute devigged probabilities using the power method.

    Parameters:
    odds (list[float]): Decimal odds for each outcome (all > 1).
    tol (float): Tolerance for root finding.
    max_iter (int): Maximum binary search iterations.

    Returns:
    list[float]: Devigged probabilities that sum to 1.
    """
    n = len(odds)
    if n == 0:
        raise ValueError("At least one odds value required")
    if any(o <= 1 for o in odds):
        raise ValueError("All decimal odds must be > 1")

    # Raw implied probabilities
    raw_probs = [1.0 / o for o in odds]
    total_raw = sum(raw_probs)

    # No vig case (very rare in practice)
    if abs(total_raw - 1.0) < 1e-12:
        return raw_probs

    # Define function f(t) = sum(1/odds_i^t) - 1
    def f(t):
        s = sum(o ** (-t) for o in odds)
        return s - 1.0

    # Find t via binary search (f is strictly decreasing)
    low, high = 0.0, 10.0

    # Expand high until f(high) < 0
    while f(high) > 0 and high < 1e6:
        high *= 2.0

    # Binary search
    for _ in range(max_iter):
        mid = (low + high) / 2.0
        if f(mid) > 0:
            low = mid
        else:
            high = mid
        if high - low < tol:
            break

    t = (low + high) / 2.0

    # Compute final probabilities
    probs = [o ** (-t) for o in odds]
    return probs


if __name__ == "__main__":
    # Example: two outcomes with odds that would yield ~ (0.3, 0.7)
    # For p=0.3 → fair odds = 1/0.3 ≈ 3.3333, for p=0.7 → 1/0.7 ≈ 1.4286
    example_odds = [3.3333, 1.4286]

    print("Power Method Devig Calculator")
    print("-" * 40)
    odds_input = input("Enter decimal odds separated by commas (e.g., 1.91,2.10): ")

    try:
        odds_list = [float(x.strip()) for x in odds_input.split(",")]
    except ValueError:
        print("Invalid input. Using example odds:", example_odds)
        odds_list = example_odds

    try:
        fair_probs = power_method_devig(odds_list)
        print("\nDevigged Probabilities:")
        for i, p in enumerate(fair_probs):
            print(f"Outcome {i+1}: {p:.6f}  ({p:.2%})")
        print(f"Sum = {sum(fair_probs):.8f}")
    except Exception as e:
        print("Error:", e)
```

```python
"""
Proportional Devigging Calculator

Given decimal odds (e.g., 1.91, 2.10), this script removes the vig
by dividing each raw implied probability (1/odds) by the sum of all
raw implied probabilities.

This corresponds to the assumption that the overround is distributed
proportionally to the size of each implied probability.
"""

def proportional_devig(odds):
    """
    Compute devigged probabilities using proportional scaling.

    Parameters:
    odds (list[float]): Decimal odds for each outcome (all > 1).

    Returns:
    list[float]: Devigged probabilities that sum to 1.
    """
    if not odds:
        raise ValueError("At least one odds value required")
    if any(o <= 1 for o in odds):
        raise ValueError("All decimal odds must be > 1")

    # Raw implied probabilities
    raw_probs = [1.0 / o for o in odds]
    total_raw = sum(raw_probs)

    # Scale to sum to 1
    fair_probs = [p / total_raw for p in raw_probs]
    return fair_probs


if __name__ == "__main__":
    print("Proportional Devigging Calculator")
    print("-" * 40)
    odds_input = input("Enter decimal odds separated by commas (e.g., 1.91,2.10): ")

    try:
        odds_list = [float(x.strip()) for x in odds_input.split(",")]
    except ValueError:
        print("Invalid input. Using example odds: 1.95, 2.05")
        odds_list = [1.95, 2.05]

    try:
        fair_probs = proportional_devig(odds_list)
        print("\nDevigged Probabilities (Proportional Method):")
        for i, p in enumerate(fair_probs):
            print(f"Outcome {i+1}: {p:.6f}  ({p:.2%})")
        print(f"Sum = {sum(fair_probs):.8f}")

        # Show original vig
        raw_sum = sum(1.0 / o for o in odds_list)
        overround = (raw_sum - 1.0) * 100
        print(f"\nOriginal overround: {overround:.2f}%")
    except Exception as e:
        print("Error:", e)
```
