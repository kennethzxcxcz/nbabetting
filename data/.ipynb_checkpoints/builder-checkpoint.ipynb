#### Data Processing and Dataset Creation

This section sets up the necessary libraries and functions to process raw basketball game data and generate a structured dataset. The dataset will be created by merging two CSV files:

1. **Schedule File** (`schedule.csv`): Contains information about games, including dates, home and away teams, and the winning team.
2. **Game Logs File** (`gamelogs.csv`): Stores detailed statistical data for each game, broken down by team.

##### Key Components:
- **Library Imports**: We use `csv` for handling CSV files, `os` for file path manipulations, and `rich` for enhanced console output with progress tracking.
- **File Path Handling**: We determine the absolute paths of the input files to ensure proper loading regardless of the working directory.
- **Data Processing**: The `create_dataset` function:
  - Reads the schedule and game log data.
  - Uses a dictionary (`defaultdict`) for efficient lookups of team statistics by date.
  - Iterates through scheduled games, retrieving corresponding statistics from the game logs.
  - Merges relevant data into a structured format and writes it to `dataset.csv`.
  - Utilizes a progress bar to track processing.

By running this section, you will generate a cleaned and structured dataset ready for analysis or machine learning applications.

```python
"""
builder.py – Merges schedule, rolling averages, and odds data into the
final model-ready dataset.csv used by model.py.

Phase 1 Upgrades
----------------
1.1  Rest Advantage : propagates days_since_last_game and is_back_to_back
     from averages.csv; computes rest_differential (home − away) as a
     derived feature.
1.2  Home/Away Splits : any *_home_avg / *_away_avg columns added by
     averager.py flow through automatically via dynamic header detection.
1.3  Leakage : find_closest_stats only looks up stats from BEFORE the
     game date (strictly less-than comparison preserved).

Dynamic Schema
--------------
Headers and row construction are driven by the actual columns present
in averages.csv, so any future features added in averager.py appear in
the dataset without requiring builder.py changes.
"""

import csv
import os
from collections import defaultdict
from datetime import datetime
from typing import Dict, List, Optional

from tqdm import tqdm


# ── Team Name Mapping ─────────────────────────────────────────────────────────

ODDS_TEAM_MAP: Dict[str, str] = {
    "Atlanta Hawks": "Atlanta",
    "Boston Celtics": "Boston",
    "Brooklyn Nets": "Brooklyn",
    "Charlotte Bobcats": "Charlotte",
    "Charlotte Hornets": "Charlotte",
    "Chicago Bulls": "Chicago",
    "Cleveland Cavaliers": "Cleveland",
    "Dallas Mavericks": "Dallas",
    "Denver Nuggets": "Denver",
    "Detroit Pistons": "Detroit",
    "Golden State Warriors": "Golden State",
    "Houston Rockets": "Houston",
    "Indiana Pacers": "Indiana",
    "Los Angeles Clippers": "LA Clippers",
    "LA Clippers": "LA Clippers",
    "Los Angeles Lakers": "LA Lakers",
    "LA Lakers": "LA Lakers",
    "Memphis Grizzlies": "Memphis",
    "Miami Heat": "Miami",
    "Milwaukee Bucks": "Milwaukee",
    "Minnesota Timberwolves": "Minnesota",
    "New Jersey Nets": "New Jersey",
    "New Orleans Hornets": "New Orleans",
    "New Orleans Pelicans": "New Orleans",
    "New York Knicks": "New York",
    "Oklahoma City Thunder": "Oklahoma City",
    "Orlando Magic": "Orlando",
    "Philadelphia 76ers": "Philadelphia",
    "Phoenix Suns": "Phoenix",
    "Portland Trail Blazers": "Portland",
    "Sacramento Kings": "Sacramento",
    "San Antonio Spurs": "San Antonio",
    "Seattle SuperSonics": "Seattle",
    "Toronto Raptors": "Toronto",
    "Utah Jazz": "Utah",
    "Washington Wizards": "Washington",
}


# ── Odds Loading ──────────────────────────────────────────────────────────────

def load_odds_data(odds_file: str) -> dict:
    """Load pre-game moneyline odds from oddsData.csv."""
    odds_dict: dict = defaultdict(dict)

    if not os.path.exists(odds_file):
        tqdm.write(
            f"Warning: Odds file '{odds_file}' not found. "
            "Moneylines will be blank."
        )
        return odds_dict

    with open(odds_file, mode="r", encoding="utf-8-sig") as fh:
        reader = csv.DictReader(fh)

        # Handle mislabelled date column ('head' vs 'date')
        date_col = "head" if "head" in (reader.fieldnames or []) else "date"
        if date_col not in (reader.fieldnames or []):
            date_col = (reader.fieldnames or [])[0] if reader.fieldnames else None
        
        if not date_col:
            tqdm.write("Warning: Could not identify date column in odds file")
            return odds_dict

        for row in reader:
            date_str = row.get(date_col, "").strip()
            season_str = row.get("season", "").strip()
            date_obj: Optional[datetime] = None

            # Standard date formats
            if "-" in date_str or "/" in date_str:
                for fmt in ("%Y-%m-%d", "%m/%d/%Y", "%m/%d/%y"):
                    try:
                        date_obj = datetime.strptime(date_str, fmt)
                        break
                    except ValueError:
                        continue

            # MMDD compact format (e.g. 1024 → Oct 24)
            if not date_obj and date_str.isdigit() and len(date_str) in (3, 4):
                date_pad = date_str.zfill(4)
                month = int(date_pad[:2])
                day = int(date_pad[2:])
                try:
                    season_yr = (
                        int(season_str[-4:])
                        if len(season_str) >= 4
                        else int(season_str)
                    )
                    year = season_yr - 1 if month >= 10 else season_yr
                    date_obj = datetime(year, month, day)
                except ValueError:
                    continue

            if not date_obj:
                continue

            team = row.get("team", "").strip()
            ml = row.get("moneyLine", "")
            opp_ml = row.get("opponentMoneyLine", "")
            hv = row.get("home/visit", "").strip().lower()

            if team:
                odds_dict[date_obj][team] = {
                    "ml": ml,
                    "opp_ml": opp_ml,
                    "hv": hv,
                }

    return odds_dict


# ── Stat Lookup ───────────────────────────────────────────────────────────────

def find_closest_stats(
    team: str,
    target_date: datetime,
    gamelogs_dict: defaultdict,
) -> Optional[Dict[str, str]]:
    """Return the most recent gamelog row for ``team`` strictly BEFORE ``target_date``."""
    gamelogs = gamelogs_dict.get(team, [])
    for date, stats in reversed(gamelogs):
        if date < target_date:
            return stats
    return None


# ── Dataset Builder ───────────────────────────────────────────────────────────

def _detect_stat_cols(averages_file: str) -> List[str]:
    """
    Peek at the header row of averages.csv and return all columns that
    are not date/team identifiers.
    """
    with open(averages_file, "r", encoding="utf-8") as fh:
        reader = csv.DictReader(fh)
        return [c for c in (reader.fieldnames or []) if c not in ("date", "team")]


def create_dataset(
    schedule_file: str,
    averages_file: str,  # FIXED: Parameter renamed for clarity
    odds_file: str,
    final_dataset_file: str,
    start_date: Optional[str] = None,
    final_date: Optional[str] = None,
) -> None:
    """Merge schedule, rolling averages, and Vegas odds into a flat dataset."""
    start_date_obj = (
        datetime.strptime(start_date, "%Y-%m-%d") if start_date else None
    )
    final_date_obj = (
        datetime.strptime(final_date, "%Y-%m-%d") if final_date else None
    )

    # ── Detect stat columns from averages.csv ─────────────────────────────────
    tqdm.write("Detecting schema from averages file...")
    stat_cols = _detect_stat_cols(averages_file)
    tqdm.write(f"   Found {len(stat_cols)} stat columns (incl. new Phase 1 features)")

    # Build dynamic headers
    headers: List[str] = (
        ["date", "home_team", "away_team", "winning_team"]
        + [f"home_{c}" for c in stat_cols]
        + ["home_moneyline"]
        + [f"away_{c}" for c in stat_cols]
        + ["away_moneyline"]
        + ["rest_differential"]  # Phase 1.1 derived feature
    )

    # ── Load Schedule ─────────────────────────────────────────────────────────
    tqdm.write("Loading schedule data...")
    schedule: List[dict] = []
    with open(schedule_file, mode="r") as fh:
        reader = csv.DictReader(fh)
        for row in reader:
            game_date = datetime.strptime(row["date"], "%Y-%m-%d")
            if start_date_obj and game_date < start_date_obj:
                continue
            if final_date_obj and game_date > final_date_obj:
                continue
            schedule.append(row)
    tqdm.write(f"   Found {len(schedule)} games in date range")

    # ── Load Gamelogs into Memory ─────────────────────────────────────────────
    tqdm.write("Loading team game logs from averages...")
    gamelogs_dict: defaultdict = defaultdict(list)
    with open(averages_file, mode="r") as fh:  # FIXED: Read from averages_file
        reader = csv.DictReader(fh)
        for row in tqdm(list(reader), desc="Loading averages", unit="row", leave=False):
            date = row["date"]
            team = row["team"]
            date_obj = datetime.strptime(date, "%Y-%m-%d")
            gamelogs_dict[team].append((date_obj, row))

    # ── Load Odds ─────────────────────────────────────────────────────────────
    tqdm.write("Loading odds dataset...")
    odds_data = load_odds_data(odds_file)

    # ── Sort by Date ──────────────────────────────────────────────────────────
    tqdm.write("Sorting team statistics by date...")
    for team in tqdm(gamelogs_dict, desc="Sorting teams", unit="team", leave=False):
        gamelogs_dict[team].sort(key=lambda x: x[0])

    # ── Build Output File ─────────────────────────────────────────────────────
    if os.path.exists(final_dataset_file):
        os.remove(final_dataset_file)

    tqdm.write("\nCreating final dataset...\n")
    rows_written: int = 0

    with open(final_dataset_file, mode="a", newline="") as fh:
        writer = csv.writer(fh)
        writer.writerow(headers)

        for game in tqdm(schedule, desc="Processing games", unit="game", colour="green"):
            date: str = game["date"]
            home_team: str = game["home_team"]
            away_team: str = game["away_team"]
            winner: str = game["winning_team"]
            game_date: datetime = datetime.strptime(date, "%Y-%m-%d")

            home_stats = find_closest_stats(home_team, game_date, gamelogs_dict)
            away_stats = find_closest_stats(away_team, game_date, gamelogs_dict)

            if not (home_stats and away_stats):
                continue

            # ── Odds Lookup ───────────────────────────────────────────────────
            matched_home = ODDS_TEAM_MAP.get(home_team)
            matched_away = ODDS_TEAM_MAP.get(away_team)
            home_ml, away_ml = "", ""

            if game_date in odds_data:
                if matched_home and matched_home in odds_data[game_date]:
                    home_ml = odds_data[game_date][matched_home]["ml"]
                if matched_away and matched_away in odds_data[game_date]:
                    away_ml = odds_data[game_date][matched_away]["ml"]
                # Failsafe: derive missing line from opponent's opp_ml
                if matched_home and not away_ml and matched_home in odds_data[game_date]:
                    away_ml = odds_data[game_date][matched_home]["opp_ml"]
                if matched_away and not home_ml and matched_away in odds_data[game_date]:
                    home_ml = odds_data[game_date][matched_away]["opp_ml"]

            # ── Phase 1.1: rest_differential ─────────────────────────────────
            try:
                home_rest = float(home_stats.get("days_since_last_game") or 7.0)
            except (TypeError, ValueError):
                home_rest = 7.0
            try:
                away_rest = float(away_stats.get("days_since_last_game") or 7.0)
            except (TypeError, ValueError):
                away_rest = 7.0
            rest_differential = round(home_rest - away_rest, 2)

            # ── Build Row Using Explicit Key Access ───────────────────────────
            home_row = [home_stats.get(c, "") for c in stat_cols] + [home_ml]
            away_row = [away_stats.get(c, "") for c in stat_cols] + [away_ml]

            row = (
                [date, home_team, away_team, winner]
                + home_row
                + away_row
                + [rest_differential]
            )
            writer.writerow(row)
            rows_written += 1

    tqdm.write(f"\nSuccessfully created dataset with {rows_written} games")
    tqdm.write(f"Saved to: {final_dataset_file}")
    tqdm.write(
        f"\nPhase 1 features in output:\n"
        f"  • home_days_since_last_game, home_is_back_to_back\n"
        f"  • away_days_since_last_game, away_is_back_to_back\n"
        f"  • rest_differential  (home − away)\n"
        f"  • home/away split averages (*_home_avg, *_away_avg) per side\n"
    )


if __name__ == "__main__":
    create_dataset(
        schedule_file="./csv/results.csv",
        averages_file="./csv/averages.csv",  # FIXED: Correct parameter
        odds_file="./csv/oddsData.csv",
        final_dataset_file="./csv/dataset.csv",
        start_date="2000-10-31",
        final_date="2025-04-13",
    )
    print("\nDone!\n")
```
